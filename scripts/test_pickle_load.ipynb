```python
# Bordas inferiores dos intervalos
border_list = np.array([0, 10, 20, 30, 40])

# Lista de ids associada aos intervalos
musicid_list = np.array(["A", "B", "C", "D"])

# Índices retornados pelo Faiss
idx_faiss = np.array([23, 5, 37])  

# 1) Encontrar os intervalos de cada índice
interval_idx = np.searchsorted(border_list, idx_faiss, side='right') - 1

# 2) Pegar o musicid correspondente a cada índice
musicid = musicid_list[interval_idx]

# 3) Pegar a posição relativa dentro do intervalo
musicid_index = idx_faiss - border_list[interval_idx]

print("interval_idx:", interval_idx)     # [2, 0, 3]
print("musicid:", musicid)               # ['C', 'A', 'D']
print("musicid_index:", musicid_index)   # [3, 5, 7]
```
